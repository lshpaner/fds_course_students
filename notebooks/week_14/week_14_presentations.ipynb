## The one idea

**You now own a loop, not a toolbox.** Question, data, explore, model, evaluate, decide, with an interval on every number and a group table on every model. Every method in this course was a way to run one step of that loop; the loop is the thing that transfers to a problem none of us has seen.

## 1. How the session runs

- Teams present in the posted order. Twelve minutes, hard stop, then three minutes of questions. One timer, visible to the presenter.
- Every team's notebook was due before the session; the talk is not the submission. If the notebook doesn't run, the rubric's automatic deduction applies regardless of how the talk goes.
- The audience is the decision-maker you named in chapter 4. Talk to them, not to me. The first two minutes and the last one should make sense to someone who has never heard the word "AUC."
- One chart per claim. If a slide has a table with more than six numbers on it, the chart that should have replaced it is missing.
- Every student fills in the peer-feedback form for every team except their own. It counts toward participation.

## 2. Peer feedback

Three lines per team, written during the three-minute question period, submitted by end of day.

1. **The decision, in your words.** Who decides, what changes. If you can't write it from the talk alone, say so; that is the most useful feedback a team can get.
2. **The number you'd act on.** The one figure from the talk that the decision-maker would take into the meeting, with its interval if they gave one. If every number had an interval, say that too.
3. **The question you didn't get to ask.** Something the evaluation, the fairness table, or the data handling left open.

Keep it specific. "Good job" and "nice slides" are not feedback.

## 3. The course in one page

| Step of the loop | What you can now do | Chapter | The habit that goes with it |
|---|---|---|---|
| **Question** | name the decision-maker, the decision, and the cost of each error before touching data | 1 | if you can't say who decides, you don't have a project yet |
| **Data** | join components on a key without losing people; name the missingness mechanism; choose and defend an imputation | 2 | write every choice down, with the assumption it makes |
| **Uncertainty** | put an interval on a proportion, a mean, a difference, an AUC; use survey weights; read a Table 1 | 3 | an estimate without an interval is a guess |
| **Explore** | univariate, bivariate, three-way; split by the strongest known predictor before believing anything | 4 | the findings log: hypotheses, not results |
| **Communicate** | the sentence first, then the chart; honest axes; small multiples; source line | 5 | would the decision-maker get it in five seconds? |
| **Explain** | regression as adjusted comparison; interactions; diagnostics; collinearity | 6 | compare crude to adjusted to see confounding |
| **Predict** | logistic regression, trees, forests, boosting; pipelines that don't leak; the accuracy trap | 7 | the baseline first; the gain has to earn the complexity |
| **Evaluate** | cutoff from a stated target; ROC and PR; DeLong; calibration; bootstrap; the rule in the decision-maker's units | 8 | the metric follows from the cost of each mistake |
| **Structure** | k-means and PCA; describe clusters before revealing the label; stability | 9 | no label, so judge by stability and usefulness |
| **Time** | decompose; autocorrelation; train on the past; beat seasonal naive; error by horizon | 10 | state the horizon with every number |
| **Relations** | centralities and what each answers; mixing; communities; spreading; what aggregation loses | 11 | pick the centrality that matches the question |
| **Text** | TF-IDF baseline; embeddings and nearest neighbors; prompting as a classifier; cost per document | 12 | an LLM is a model and gets a test set |
| **Fairness** | per-group metrics with intervals; the impossibility result; group cutoffs; who bears the cost | 13 | every model ships with a group table |

The libraries, for reference: pandas and scikit-learn throughout; statsmodels (chapters 3, 6, 10); `eda_toolkit` (1 to 5); `model_tuner` and `model_metrics` (6 to 8); networkx (11); `equiboots` (13).

## 4. What to do next

Three directions, depending on what you found you liked.

**If you liked the modeling.** ISLR cover to cover, then its deep-learning chapter; Hastie, Tibshirani & Friedman, *The Elements of Statistical Learning*, for the proofs behind chapters 6 to 9. Kaggle for practice, with the rule that you write the baseline before you read the forums.

**If you liked the inference.** Gelman, Hill & Vehtari, *Regression and Other Stories*. McElreath, *Statistical Rethinking*, for the Bayesian version of everything in chapter 3 and 6. Hernán & Robins, *Causal Inference: What If*, for the question chapter 4 kept raising and never answered: when does "adjusted for" mean "caused by"?

**If you liked the decisions.** Provost & Fawcett, *Data Science for Business*, for the cost-sensitive framing of chapter 8 at book length. Barocas, Hardt & Narayanan, *Fairness and Machine Learning*, in full. And the hardest skill: write a one-page memo for a real decision-maker, with one chart and one number with an interval, and watch what they do with it.
